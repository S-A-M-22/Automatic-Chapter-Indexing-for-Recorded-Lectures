VIDEO -> TIMESTAMP-BASED TRANSCRIPT (CODE)

```python
# =========================================================================================
# STEP 5: AUDIO EXTRACTION FUNCTIONS
# =========================================================================================

def extract_audio_from_video(video_path: str, audio_path: str, sample_rate: int = 16000) -> bool:
    """
    Extract audio from video file using FFmpeg.
    
    Args:
        video_path: Path to input video file
        audio_path: Path to output audio file
        sample_rate: Audio sample rate (16kHz is optimal for Whisper)
    
    Returns:
        True if successful, False otherwise
    """
    try:
        cmd = [
            'ffmpeg',
            '-i', video_path,
            '-vn',  # Disable video
            '-acodec', 'pcm_s16le',  # 16-bit PCM
            '-ar', str(sample_rate),  # Sample rate
            '-ac', '1',  # Mono
            '-y',  # Overwrite output file
            audio_path
        ]
        
        result = subprocess.run(
            cmd,
            capture_output=True,
            text=True,
            timeout=3600  # 1 hour timeout for long videos
        )
        
        if result.returncode == 0:
            return True
        else:
            print(f"❌ FFmpeg error: {result.stderr}")
            return False
            
    except subprocess.TimeoutExpired:
        print("❌ Audio extraction timed out (video too long?)")
        return False
    except Exception as e:
        print(f"❌ Error extracting audio: {e}")
        return False

# =========================================================================================
# STEP 6: TRANSCRIPTION FUNCTIONS
# =========================================================================================

def transcribe_audio_with_timestamps(
    audio_path: str,
    model: whisper.Whisper,
    language: str = "en"
) -> Dict:
    """
    Transcribe audio file with word-level timestamps.
    
    Args:
        audio_path: Path to audio file
        model: Loaded Whisper model
        language: Language code (e.g., 'en' for English)
    
    Returns:
        Dictionary containing transcription results with segments
    """
    try:
        print(f"  Transcribing audio (this may take several minutes)...")
        
        # Transcribe with word timestamps
        result = model.transcribe(
            audio_path,
            language=language,
            verbose=False,
            word_timestamps=True,
            task="transcribe"
        )
        
        return result
        
    except Exception as e:
        print(f"❌ Error during transcription: {e}")
        return None

# =========================================================================================
# STEP 7: FORMAT OUTPUT FUNCTIONS
# =========================================================================================

def format_timestamp(seconds: float) -> str:
    """Convert seconds to HH:MM:SS.mmm format."""
    hours = int(seconds // 3600)
    minutes = int((seconds % 3600) // 60)
    secs = seconds % 60
    return f"{hours:02d}:{minutes:02d}:{secs:06.3f}"

def save_transcript(result: Dict, output_path: str, include_word_timestamps: bool = False):
    """
    Save transcription results to file.
    
    Args:
        result: Whisper transcription result
        output_path: Path to output text file
        include_word_timestamps: If True, include word-level timestamps
    """
    try:
        with open(output_path, 'w', encoding='utf-8') as f:
            # Write full transcript
            f.write("=" * 80 + "\n")
            f.write("FULL TRANSCRIPT\n")
            f.write("=" * 80 + "\n\n")
            f.write(result['text'].strip() + "\n\n")
            
            # Write segment-level timestamps
            f.write("=" * 80 + "\n")
            f.write("TIMESTAMPED SEGMENTS\n")
            f.write("=" * 80 + "\n\n")
            
            for segment in result['segments']:
                start = format_timestamp(segment['start'])
                end = format_timestamp(segment['end'])
                text = segment['text'].strip()
                
                f.write(f"[{start} --> {end}]\n")
                f.write(f"{text}\n\n")
                
                # Optionally include word-level timestamps
                if include_word_timestamps and 'words' in segment:
                    f.write("  Word-level timestamps:\n")
                    for word_info in segment['words']:
                        word_start = format_timestamp(word_info['start'])
                        word_end = format_timestamp(word_info['end'])
                        word = word_info['word'].strip()
                        f.write(f"    [{word_start} - {word_end}] {word}\n")
                    f.write("\n")
        
        print(f"✅ Transcript saved: {output_path}")
        return True
        
    except Exception as e:
        print(f"❌ Error saving transcript: {e}")
        return False

# =========================================================================================
# STEP 8: MAIN PROCESSING FUNCTION
# =========================================================================================

def process_video(
    video_path: str,
    output_dir: str,
    model: whisper.Whisper,
    language: str = "en",
    include_word_timestamps: bool = False
) -> bool:
    """
    Complete pipeline: extract audio, transcribe, save transcript.
    
    Args:
        video_path: Path to input video file
        output_dir: DIRECTORY to save outputs
        model: Loaded Whisper model
        language: Language code
        include_word_timestamps: Include word-level timestamps
    
    Returns:
        True if successful, False otherwise
    """
    video_path = Path(video_path)
    output_dir = Path(output_dir)
    
    if not video_path.exists():
        print(f"❌ Video file not found: {video_path}")
        return False
    
    # Create output DIRECTORY
    output_dir.mkdir(parents=True, exist_ok=True)
    
    # Generate output filenames
    base_name = video_path.stem
    audio_path = output_dir / f"{base_name}_audio.wav"
    transcript_path = output_dir / f"{base_name}_transcript.txt"
    
    print(f"\n{'='*80}")
    print(f"Processing: {video_path.name}")
    print(f"{'='*80}")
    
    # Step 1: Extract audio
    print("Step 1/3: Extracting audio from video...")
    if not extract_audio_from_video(str(video_path), str(audio_path)):
        return False
    print(f"✅ Audio extracted: {audio_path}")
    
    # Step 2: Transcribe
    print("Step 2/3: Transcribing audio with Whisper AI...")
    result = transcribe_audio_with_timestamps(str(audio_path), model, language)
    if result is None:
        return False
    print(f"✅ Transcription complete")
    
    # Step 3: Save transcript
    print("Step 3/3: Saving transcript...")
    success = save_transcript(result, str(transcript_path), include_word_timestamps)
    
    # Cleanup: Remove temporary audio file
    try:
        audio_path.unlink()
        print(f"✅ Cleaned up temporary audio file")
    except Exception as e:
        print(f"⚠️  Could not delete temporary audio file: {e}")
    
    return success

# =========================================================================================
# STEP 9: BATCH PROCESSING
# =========================================================================================

def process_directory(
    input_dir: str,
    output_dir: str,
    model: whisper.Whisper,
    language: str = "en",
    include_word_timestamps: bool = False,
    video_extensions: List[str] = ['.mp4', '.mov', '.avi', '.mkv']
) -> Tuple[int, int]:
    """
    Process all video files in a directory.
    
    Args:
        input_dir: Directory containing video files
        output_dir: Directory to save transcripts
        model: Loaded Whisper model
        language: Language code
        include_word_timestamps: Include word-level timestamps
        video_extensions: List of video file extensions to process
    
    Returns:
        Tuple of (successful_count, failed_count)
    """
    input_path = Path(input_dir)
    
    if not input_path.exists():
        print(f"❌ Input directory not found: {input_dir}")
        return 0, 0
    
    # Find all video files
    video_files = []
    for ext in video_extensions:
        video_files.extend(input_path.glob(f"*{ext}"))
        video_files.extend(input_path.glob(f"*{ext.upper()}"))
    
    if not video_files:
        print(f"❌ No video files found in: {input_dir}")
        return 0, 0
    
    print(f"\n{'='*80}")
    print(f"Found {len(video_files)} video file(s) to process")
    print(f"{'='*80}")
    
    successful = 0
    failed = 0
    
    for video_file in video_files:
        try:
            if process_video(
                str(video_file),
                output_dir,
                model,
                language,
                include_word_timestamps
            ):
                successful += 1
            else:
                failed += 1
        except Exception as e:
            print(f"❌ Unexpected error processing {video_file.name}: {e}")
            failed += 1
    
    return successful, failed

# =========================================================================================
# STEP 10: RUN THE PIPELINE
# =========================================================================================

if __name__ == "__main__":
    # Configuration
    INPUT_DIRECTORY = "Training Data"
    OUTPUT_DIRECTORY = "Transcripts"
    LANGUAGE = "en"  # Change to your video language
    INCLUDE_WORD_TIMESTAMPS = False  # Set to True for word-level timestamps
    
    print("\n" + "="*80)
    print("VIDEO TRANSCRIPT EXTRACTION PIPELINE")
    print("="*80)
    print(f"Input directory:  {INPUT_DIRECTORY}")
    print(f"Output directory: {OUTPUT_DIRECTORY}")
    print(f"Language:         {LANGUAGE}")
    print(f"Model:            {MODEL_SIZE}")
    print("="*80 + "\n")
    
    # Process all videos
    successful, failed = process_directory(
        INPUT_DIRECTORY,
        OUTPUT_DIRECTORY,
        model,
        LANGUAGE,
        INCLUDE_WORD_TIMESTAMPS
    )
    
    # Summary
    print("\n" + "="*80)
    print("PROCESSING COMPLETE")
    print("="*80)
    print(f"✅ Successful: {successful}")
    print(f"❌ Failed:     {failed}")
    print(f"📁 Output:     {OUTPUT_DIRECTORY}")
    print("="*80 + "\n")
```

LOAD & CLEAN TRANSCRIPT

```python
from numpy import full
import datetime

FILES = []
DIRECTORY = "Transcripts"
EXTRACTED_SEGMENTS = []

# Check the Directory location exists or not and add files to the FILES list
for files in os.listdir(DIRECTORY):
    file_path = os.path.join(DIRECTORY, files)
    if os.path.isfile(file_path):
        FILES.append(file_path)
    else:
        raise Exception("File not found: ", file_path)
    
    
"""
FUNCTION: To clean the filler words from the transcript
INPUT: str (i.e. text from transcript)
OUTPUT: str (cleaned text)
"""
def clean_filler_words(line: str) -> str:
    # List of filler word upto the Developer's knowledge
    filler_words = ["um", "uh...", "um...", "uh", "...", "like", "you know", "so", "actually", "basically", "right", "cool", "I mean"]
    line = line.lower()
    for word in filler_words:
        line = line.replace(word+" ", "")
    return line.strip()


"""
FUNCTION: To load transcript from file and return a list of format: [full_transcript, (start_time, end_time, cleaned_text)]
INPUT: str (i.e. File path to the location of transcript)
OUTPUT: List[str, Tuple[datetime, datetime, str]]
"""
def load_transcript_from_file(file_path: str) -> list[str, Tuple[datetime, datetime, str]]:
    with open(file_path, "r+") as whisper_transcript:
        read_text = whisper_transcript.readlines()
        full_transcript = ""
        nested_list = []
        # print(lines)
        
        line = ""
        print(len(read_text))
        index = 0
        while index >= 10 and index < len(read_text):
            start_time = float(read_text[index].split(' --> ')[0])
            start_datetime = convert_timestamp_to_datetime(start_time)
            end_time = float(read_text[index].split(' --> ')[1])
            end_datetime = convert_timestamp_to_datetime(end_time)
            # TODO: Convert the timestamp to datetime format(?)
            text = " ".join(read_text[index+1])
            cleaned_text = clean_filler_words(text)
            nested_list.append((start_datetime, end_datetime, cleaned_text))
            
            if cleaned_text[-1] == "." or cleaned_text[-1] == "?" or cleaned_text[-1] == "!":
                full_transcript += cleaned_text + "\n"
            else:
                full_transcript += cleaned_text + " "
            
            index += 3
    
    return full_transcript, nested_list

file_path = "Transcripts/Week 01 - Embedded S_transcript.txt"
EXTRACTED_SEGMENTS = load_transcript_from_file(file_path)


"""
FUNCTION: Convert HH:MM:SS.mmm timestamp to datetime format
INPUT: str (Timestamp = HH:MM:SS.mmm)
OUTPUT: Datetime object
"""
def convert_timestamp_to_datetime(timestamp: str) -> datetime:
    from datetime import datetime
    return datetime.strptime(timestamp, "%H:%M:%S.%f")
```
